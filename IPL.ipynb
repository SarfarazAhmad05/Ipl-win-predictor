```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# 1. Clean NaN and inf values
X_train_clean = X_train.copy()
X_test_clean = X_test.copy()

X_train_clean['city'] = X_train_clean['city'].fillna('Unknown')
X_test_clean['city'] = X_test_clean['city'].fillna('Unknown')

for col in ['crr', 'rrr']:
    X_train_clean[col] = X_train_clean[col].replace([np.inf, -np.inf], np.nan)
    X_test_clean[col] = X_test_clean[col].replace([np.inf, -np.inf], np.nan)
    med = X_train_clean[col].median()
    X_train_clean[col] = X_train_clean[col].fillna(med)
    X_test_clean[col] = X_test_clean[col].fillna(med)

# 2. Build and fit the pipeline
pipe = Pipeline(steps=[
    ('step1', trf),
    ('step2', LogisticRegression(solver='liblinear'))
])

pipe.fit(X_train_clean, y_train)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder

cat_cols = ['batting_team', 'bowling_team', 'city']
num_cols = ['runs_left', 'balls_left', 'wickets_left', 'runs_total_x', 'crr', 'rrr']

cat_pipe = Pipeline([
    ('impute', SimpleImputer(strategy='constant', fill_value='Unknown')),
    ('ohe', OneHotEncoder(sparse_output=False, drop='first', handle_unknown='ignore'))
])

num_pipe = Pipeline([
    ('impute', SimpleImputer(strategy='median'))
])

trf = ColumnTransformer([
    ('cat', cat_pipe, cat_cols),
    ('num', num_pipe, num_cols)
])

pipe = Pipeline([
    ('step1', trf),
    ('step2', LogisticRegression(solver='liblinear'))
])

# Replace +/-inf with NaN first so SimpleImputer handles them cleanly
X_train_inf = X_train.replace([np.inf, -np.inf], np.nan)
X_test_inf = X_test.replace([np.inf, -np.inf], np.nan)

pipe.fit(X_train_inf, y_train)
y_pred = pipe.predict(X_test_inf)
```

```python
import numpy as np
import pandas as pd

def match_progression(delivery_df, match_id, pipe):
    df = delivery_df[delivery_df['match_id'] == match_id].copy()

    # Filter to 2nd innings chase if present
    innings_col = 'innings' if 'innings' in df.columns else ('innings_y' if 'innings_y' in df.columns else None)
    if innings_col is not None and (df[innings_col] == 2).any():
        df = df[df[innings_col] == 2].copy()

    # Keep end-of-over deliveries when available
    if 'ball' in df.columns and (df['ball'] == 6).any():
        df = df[df['ball'] == 6].copy()

    target = int(df['runs_total_x'].iloc[0]) if len(df) > 0 and 'runs_total_x' in df.columns else None
    print(f"Target- {target}")

    feature_cols = [
        'batting_team', 'bowling_team', 'city',
        'runs_left', 'balls_left', 'wickets_left',
        'runs_total_x', 'crr', 'rrr'
    ]
    X_match = df[feature_cols].replace([np.inf, -np.inf], np.nan).copy()
    X_match['city'] = X_match['city'].fillna('Unknown')

    probs = pipe.predict_proba(X_match)

    temp_df = pd.DataFrame({
        'end_of_over': df['over'].values,
        'runs_left': df['runs_left'].values,
        'balls_left': df['balls_left'].values,
        'wickets_left': df['wickets_left'].values,
        'lose_prob': np.round(probs[:, 0] * 100, 1),
        'win_prob': np.round(probs[:, 1] * 100, 1),
    })

    return temp_df, target
```

```python
import numpy as np
import pandas as pd

def match_progression(x_df, match_id, pipe):
    match = x_df[x_df['match_id'] == match_id].copy()
    match = match[match['ball'] == 6].copy()

    cols_needed = [
        'batting_team', 'bowling_team', 'city',
        'runs_left', 'balls_left', 'wickets_left',
        'runs_total_x', 'crr', 'rrr'
    ]
    temp_df = match[cols_needed].replace([np.inf, -np.inf], np.nan).dropna().copy()
    temp_df = temp_df[temp_df['balls_left'] != 0].copy()

    if temp_df.empty:
        print(f"No end-of-over rows found for match_id={match_id}.")
        return pd.DataFrame(), None

    result = pipe.predict_proba(temp_df)
    temp_df['lose'] = np.round(result.T[0] * 100, 1)
    temp_df['win'] = np.round(result.T[1] * 100, 1)
    temp_df['end_of_over'] = range(1, temp_df.shape[0] + 1)

    target = temp_df['runs_total_x'].values[0]

    runs = list(temp_df['runs_left'].values)
    new_runs = runs[:]
    runs.insert(0, target)
    temp_df['runs_after_over'] = np.array(runs)[:-1] - np.array(new_runs)

    wickets = list(temp_df['wickets_left'].values)
    new_wickets = wickets[:]
    new_wickets.insert(0, 10)
    wickets.append(0)
    w = np.array(wickets)
    nw = np.array(new_wickets)
    temp_df['wickets_in_over'] = (nw - w)[0:temp_df.shape[0]]

    print("Target-", target)
    temp_df = temp_df[['end_of_over', 'runs_after_over', 'wickets_in_over', 'lose', 'win']]
    return temp_df, target
```
